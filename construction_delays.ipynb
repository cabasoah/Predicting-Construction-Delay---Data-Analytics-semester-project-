```python
# Formatting & Standardization - We converted the date columns from string to datetime
df['Start_Date'] = pd.to_datetime(df['Start_Date'])
df['End_Date'] = pd.to_datetime(df['End_Date'])
print('Converted Start_Date and End_Date to datetime.')
print('Confirming date data types change:')
print(df.dtypes)
```

```python
# Visualizing numeric features
numeric_columns = df.select_dtypes(include=[np.number]).columns.tolist()
print('Numeric columns:', numeric_columns)

for col in numeric_columns:
    plt.figure(figsize=(8, 4))
    sns.histplot(df[col].dropna(), kde=True, bins=30)
    plt.title(f'Histogram of {col}')
    plt.tight_layout()
    plt.show()

# If there are 4 or more numeric columns, plot a correlation heatmap
if len(numeric_columns) >= 4:
    numeric_df = df.select_dtypes(include=[np.number])
    plt.figure(figsize=(12, 10))
    sns.heatmap(numeric_df.corr(), annot=True, fmt='.2f', cmap='coolwarm')
    plt.title('Correlation Heatmap for Numeric Variables')
    plt.tight_layout()
    plt.show()
```

```python
# Visualize categorical features

# List of categorical features
categorical_features = ['Project_Type', 'Location', 'Weather_Condition', 'Anomaly_Detected', 'Risk_Level']

# Loop through each categorical column and plot count distribution
for col in categorical_features:
    plt.figure(figsize=(8, 4))
    sns.countplot(y=col, data=df, order=df[col].value_counts().index)
    plt.title(f'Distribution of {col}')
    plt.tight_layout()
    plt.show()
```

```python
#Outliers Visualization
#ploting all numeric columns
for col in df.select_dtypes(include='number').columns:
    plt.figure(figsize=(6, 2))
    sns.boxplot(x=df[col])
    plt.title(f"Boxplot of {col}")
    plt.show()
```

```python
# Categorical Features Visualization

# List of categorical features
categorical_features = ['Project_Type', 'Location', 'Weather_Condition', 'Anomaly_Detected', 'Risk_Level']

for col in categorical_features:
    plt.figure(figsize=(6, 2))
    sns.countplot(x=col, data=df, order=df[col].value_counts().index)
    plt.title(f"Count Plot of {col}")
    plt.xticks(rotation=45, ha='right')
    plt.tight_layout()
    plt.show()
```

```python
# We created a copy of the original dataframe for our modeling
df_model = df.copy()
```

```python
# We dropped these columns because they are not directly useful for our prediction

# Columns to drop
cols_to_drop = [
    'Project_ID',
    'End_Date',
    'Actual_Cost',
    'Cost_Overrun',
    'Actual_Duration',
    'Schedule_Deviation',
    'Completion_Percentage',
    'Risk_Level' 
]

# Drop them from the modeling DataFrame
df_model.drop(columns=cols_to_drop, inplace=True, errors='ignore')

print('Dropped columns:', cols_to_drop)
```

```python
# One-hot encoding
categorical_cols = ['Project_Type', 'Location', 'Weather_Condition', 'Anomaly_Detected', 'Image_Analysis_Score']

# One-hot encode categorical features
df_model = pd.get_dummies(df_model, columns=categorical_cols, drop_first=False)

print("Encoding complete. New shape:", df_model.shape)
```

```python
# Feature Scaling
from sklearn.preprocessing import RobustScaler

numeric_cols_to_scale = [
    'Planned_Cost', 'Planned_Duration', 'Vibration_Level', 
    'Crack_Width', 'Load_Bearing_Capacity', 'Energy_Consumption', 
    'Material_Usage', 'Labor_Hours'
]

scaler = RobustScaler()
df_model[numeric_cols_to_scale] = scaler.fit_transform(df_model[numeric_cols_to_scale])
```

```python
# Confirming if our dataset is clean after preprocessing
print(df_model.describe().T)
```

```python
#Checking categorical encoding
print(df_model.filter(like='_').head())
print("==============================================================")
#checking duplicate
print(df_model.duplicated().sum())
print("==============================================================")
#checking datatypes
print(df_model.dtypes)
```

```python
# Save the clean dataset to a CSV file
df_model.to_csv("clean_construction_data.csv", index=False)
print("Clean dataset saved as 'clean_construction_data.csv'")
```
